# 📊 Student Performance Analysis
## Step 1: Exploratory Data Analysis (EDA)

This notebook performs an in-depth exploratory data analysis on the Student Performance dataset.
We will examine distributions, correlations, and key insights from the data.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

print('Libraries imported successfully!')
```

## 2. Load the Dataset

```python
df = pd.read_csv('Student_Performance_Data.csv')
print('Dataset loaded successfully!')
print(f'Shape: {df.shape}')
df.head(10)
```

## 3. Dataset Overview

```python
print('=== Dataset Info ===')
df.info()
```

```python
print('=== Descriptive Statistics ===')
df.describe().round(2)
```

```python
print('=== Missing Values ===')
missing = df.isnull().sum()
print(missing)
print(f'\nTotal missing values: {missing.sum()}')
```

## 4. Distribution of Each Feature

```python
features = ['Study_Hours', 'Attendance_Percentage', 'Scores']
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

colors = ['#4C72B0', '#55A868', '#C44E52']

for i, col in enumerate(features):
    axes[i].hist(df[col], bins=20, color=colors[i], edgecolor='white', alpha=0.85)
    axes[i].set_title(f'Distribution of {col}', fontsize=13, fontweight='bold')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frequency')

plt.suptitle('Feature Distributions', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## 5. Box Plots — Detecting Outliers

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, col in enumerate(features):
    sns.boxplot(y=df[col], ax=axes[i], color=colors[i])
    axes[i].set_title(f'Box Plot: {col}', fontsize=13, fontweight='bold')

plt.suptitle('Outlier Detection via Box Plots', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## 6. Correlation Analysis

```python
corr_matrix = df[features].corr()
print('=== Correlation Matrix ===')
print(corr_matrix.round(3))

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, fmt='.3f', cmap='coolwarm',
            linewidths=0.5, square=True, cbar_kws={'shrink': 0.8})
plt.title('Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 7. Scatter Plots — Relationships with Score

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.scatterplot(data=df, x='Study_Hours', y='Scores', ax=axes[0],
                color='#4C72B0', alpha=0.7, edgecolor='white')
axes[0].set_title('Study Hours vs Scores', fontsize=13, fontweight='bold')

sns.scatterplot(data=df, x='Attendance_Percentage', y='Scores', ax=axes[1],
                color='#55A868', alpha=0.7, edgecolor='white')
axes[1].set_title('Attendance Percentage vs Scores', fontsize=13, fontweight='bold')

plt.suptitle('Feature vs Score Relationships', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## 8. Pair Plot

```python
sns.pairplot(df[features], diag_kind='kde', plot_kws={'alpha': 0.6})
plt.suptitle('Pair Plot — Student Performance Features', y=1.02, fontsize=14, fontweight='bold')
plt.show()
```

## 9. Key Insights Summary

```python
print('=' * 55)
print('        KEY INSIGHTS FROM EDA')
print('=' * 55)
print(f"  Total Students         : {len(df)}")
print(f"  Avg Study Hours        : {df['Study_Hours'].mean():.2f} hrs")
print(f"  Avg Attendance         : {df['Attendance_Percentage'].mean():.2f} %")
print(f"  Avg Score              : {df['Scores'].mean():.2f}")
print(f"  Min Score              : {df['Scores'].min():.2f}")
print(f"  Max Score              : {df['Scores'].max():.2f}")
print(f"  Corr (Study vs Score)  : {df['Study_Hours'].corr(df['Scores']):.3f}")
print(f"  Corr (Attend vs Score) : {df['Attendance_Percentage'].corr(df['Scores']):.3f}")
print('=' * 55)
```
